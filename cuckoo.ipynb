```python

import numpy as np

# Parameters
N = 10                  # Number of nests
ITERATIONS = 100
PA = 0.25               # Nest abandonment probability
ALPHA = 0.01            # Step size
BETA = 1.5              # Levy flight parameter

# Engineering objective function
# Minimize f(x) = x1^2 + x2^2
def fitness(x):
    return x[0]**2 + x[1]**2

# Generate initial nests
nests = np.random.uniform(-10, 10, (N, 2))

# Evaluate initial fitness
scores = np.array([fitness(x) for x in nests])

# Find initial best solution
best_index = np.argmin(scores)
best_nest = nests[best_index].copy()
best_fitness = scores[best_index]

# Cuckoo Search iterations
for iteration in range(ITERATIONS):

    for i in range(N):

        # Generate new solution using Levy flight
        u = np.random.normal(0, 1, 2)
        v = np.random.normal(0, 1, 2)

        step = u / (np.abs(v)**(1 / BETA) + 1e-10)

        new_nest = nests[i] + ALPHA * step
        new_nest = np.clip(new_nest, -10, 10)

        # Evaluate new solution
        new_fitness = fitness(new_nest)

        # Select a random existing nest
        j = np.random.randint(0, N)

        # Replace nest if new solution is better
        if new_fitness < scores[j]:
            nests[j] = new_nest.copy()
            scores[j] = new_fitness

        # Update global best solution
        if new_fitness < best_fitness:
            best_fitness = new_fitness
            best_nest = new_nest.copy()

    # Abandon a fraction of poor nests
    for i in range(N):
        if np.random.rand() < PA:
            nests[i] = np.random.uniform(-10, 10, 2)
            scores[i] = fitness(nests[i])

            # Update best solution
            if scores[i] < best_fitness:
                best_fitness = scores[i]
                best_nest = nests[i].copy()

    print(
        f"Iteration {iteration + 1}: "
        f"Best Fitness = {best_fitness:.6f}"
    )

# Final output
print("\n--- BEST SOLUTION ---")
print("Best Nest:", best_nest)
print("Best Fitness:", best_fitness)
```

```python

import numpy as np
import random

# Number of cities, ants and iterations
NUM_CITIES = 10
NUM_ANTS = 20
ITERATIONS = 100

# ACO parameters
ALPHA = 1.0
BETA = 2.0
EVAPORATION = 0.5
Q = 100

# Distance matrix between 10 cities
distance = np.array([
    [0, 12, 10, 19, 8, 15, 14, 20, 11, 17],
    [12, 0, 3, 7, 2, 9, 8, 13, 6, 10],
    [10, 3, 0, 6, 5, 8, 7, 12, 4, 9],
    [19, 7, 6, 0, 9, 5, 4, 8, 11, 6],
    [8, 2, 5, 9, 0, 10, 9, 14, 7, 12],
    [15, 9, 8, 5, 10, 0, 3, 6, 9, 4],
    [14, 8, 7, 4, 9, 3, 0, 5, 8, 2],
    [20, 13, 12, 8, 14, 6, 5, 0, 10, 7],
    [11, 6, 4, 11, 7, 9, 8, 10, 0, 5],
    [17, 10, 9, 6, 12, 4, 2, 7, 5, 0]
], dtype=float)

# Initialize pheromone and heuristic matrices
pheromone = np.ones((NUM_CITIES, NUM_CITIES))
heuristic = 1 / (distance + np.eye(NUM_CITIES))

# Store the best route
best_route = None
best_distance = float('inf')

# ACO algorithm
for iteration in range(ITERATIONS):

    all_routes = []
    all_distances = []

    for ant in range(NUM_ANTS):

        # Select random starting city
        route = [random.randint(0, NUM_CITIES - 1)]

        # Construct a complete route
        while len(route) < NUM_CITIES:

            current = route[-1]
            unvisited = [
                city for city in range(NUM_CITIES)
                if city not in route
            ]

            # Calculate transition probabilities
            probabilities = [
                (pheromone[current][city] ** ALPHA) *
                (heuristic[current][city] ** BETA)
                for city in unvisited
            ]

            total = sum(probabilities)
            probabilities = [p / total for p in probabilities]

            next_city = random.choices(
                unvisited, weights=probabilities, k=1
            )[0]

            route.append(next_city)

        # Calculate total distance including return to start
        total_distance = sum(
            distance[route[i]][route[i + 1]]
            for i in range(NUM_CITIES - 1)
        )

        total_distance += distance[route[-1]][route[0]]

        all_routes.append(route)
        all_distances.append(total_distance)

        # Update global best route
        if total_distance < best_distance:
            best_distance = total_distance
            best_route = route.copy()

    # Pheromone evaporation
    pheromone *= (1 - EVAPORATION)

    # Pheromone deposition based on route quality
    for route, route_distance in zip(all_routes, all_distances):

        deposit = Q / route_distance

        for i in range(NUM_CITIES - 1):
            a, b = route[i], route[i + 1]
            pheromone[a][b] += deposit
            pheromone[b][a] += deposit

        # Include return edge
        a, b = route[-1], route[0]
        pheromone[a][b] += deposit
        pheromone[b][a] += deposit

    print(
        f"Iteration {iteration + 1}: "
        f"Best Distance = {best_distance:.2f}"
    )

# Final result
print("\n----- FINAL BEST ROUTE -----")
print("Best Route:", [city + 1 for city in best_route])
print("Shortest Distance:", best_distance)
```
